## Step 3 — Apni Env Values Bharo

Neeche har quotes `""` ke beech apni value daalo (jaise Render pe env vars daalte ho). Jo optional hai use khaali chhod sakte ho, baaki zaroori hain.

```
import os

# ── Zaroori values ──
os.environ["API_ID"]     = ""   # my.telegram.org se milta hai
os.environ["API_HASH"]   = ""   # my.telegram.org se milta hai
os.environ["BOT_TOKEN"]  = ""   # BotFather se
os.environ["MONGO_DB"]   = ""   # MongoDB Atlas connection string
os.environ["OWNER_ID"]   = ""   # tumhari Telegram numeric user id (space-separated agar multiple)

# ── Optional (Render pe already set hain to yahan bhi wahi daalo) ──
os.environ["DB_NAME"]    = "telegram_downloader"
os.environ["LOG_GROUP"]  = ""
os.environ["FORCE_SUB"]  = ""
os.environ["MASTER_KEY"] = ""
os.environ["IV_KEY"]     = ""

print("Env values set.")
```

## Step 4 — Bot Ko Background Me Chalao

Ye cell bot ko background process ki tarah start karega taaki notebook block na ho aur tum neeche wale cell se logs dekh sako.

```
import subprocess, os

log_file = open("bot_log.txt", "w")
proc = subprocess.Popen(
    ["python", "main.py"],
    stdout=log_file,
    stderr=subprocess.STDOUT,
    env=os.environ.copy(),
)
print(f"Bot started with PID {proc.pid}. Neeche wale cell se logs check karo.")
```

## Step 6 — Session Zinda Rakhna (Keep-Alive Loop)

Ye cell chalte rehne se Colab runtime ko notebook 'busy' dikhta rahega. Ise **run karke chhod do**, ise stop mat karo jab tak bot band nahi karna. Isse tab/browser ka connection bana rehna chahiye, disconnect abhi bhi ho sakta hai (Colab ki apni time-limit ke wajah se).

```
import time
print("Keep-alive shuru. Ise mat roko — bot chalta rahega jab tak ye cell chal rahi hai.")
while True:
    time.sleep(60)
```

## Bot Band Karne Ke Liye

Upar wali keep-alive cell ko stop karo (■ button), phir ye cell run karo.

```
try:
    proc.terminate()
    print("Bot stopped.")
except Exception as e:
    print(f"Already stopped or error: {e}")
```
